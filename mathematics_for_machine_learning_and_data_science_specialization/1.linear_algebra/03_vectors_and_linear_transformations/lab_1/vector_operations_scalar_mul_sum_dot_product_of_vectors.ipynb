# **Vector Operations: Scalar Multiplication, Sum and Dot Product of Vectors**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_vectors(list_v, list_label, list_color):
    
    # plt.subplots() creates a figure (the window/canvas) and one or more axes (the plotting area) all at once.
    _ , ax = plt.subplots(figsize=(10, 10))
    
    # tick_params() control the resolution of the axis
    ax.tick_params(axis='x', labelsize=14)
    ax.tick_params(axis='y', labelsize=14)
    
    # Controls POSITIONS (where ticks appear)
    ax.set_xticks(np.arange(-10, 10))
    ax.set_yticks(np.arange(-10, 10))
    
    # Controls VIEWING WINDOW (what range you see)
    plt.axis([-10, 10, -10, 10])
    
    for i, v in enumerate(list_v):
        # This line is purely about label positioning, to just avoid putting the label on the top of the vector.
        sgn = 0.4 * np.array([[1] if i==0 else [i] for i in np.sign(v)])
    
        # angles='xy': you're telling matplotlib: "Calculate the direction of this arrow using my data coordinate system, 
        # then figure out how to render it on screen correctly." 
        # scale_units='xy', it prevents matplot lib to draw the vector according to screen
        # quiver is used to draw the arrow
        plt.quiver(v[0], v[1], color=list_color[i], angles='xy', scale_units='xy', scale=1)
        ax.text(v[0]-0.2+sgn[0], v[1]-0.2+sgn[1], list_label[i], fontsize=14, color=list_color[i])
    
    # To make the diagram grid
    plt.grid()
    
    plt.gca().set_aspect("equal")
    plt.show
```

```python
v = np.array([[1],[3]])
# Arguments: list of vectors as NumPy arrays, labels, colors.
plot_vectors([v], [f"$v$"], ["black"])
```

```python
# Scalar Multiplication
plot_vectors([v, 2*v, -2*v], [f"$v$", f"$2v$", f"$-2v$"], ["black", "green", "blue"])
```

```python
# Sum of Vectors
v = np.array([[1],[3]])
w = np.array([[4],[-1]])

plot_vectors([v, w, v + w], [f"$v$", f"$w$", f"$v + w$"], ["black", "black", "red"])
```

```python
# Norm of a vector
v = np.array([[1],[3]])
print("Norm of a vector v is", np.linalg.norm(v))
```

```python
# dot function to calculate the dot product
def dot (x, y):
    sum = 0
    for i, j in zip(x, y):
        sum += (i * j)

    return sum
```

```python
x = [1, -2, -5]
y = [4, 3, -1]

print("The dot product of x and y is", dot(x, y))
```

```python
# NumPy has its internal dot product
x = [1, -2, -5]
y = [4, 3, -1]

# Note that x and y are not NumPy Arrays
print("np.dot(x,y) function returns dot product of x and y:", np.dot(x, y)) 
```

```python
# @ symbol refer to dot product operations, only can be used with NumPy Arrays
print("This line output is a dot product of x and y: ", np.array(x) @ np.array(y))

print("\nThis line output is an error:")
try:
    print(x @ y)
except TypeError as err:
    print(err)
```

```python
# Measure of different methods to calculate dot product
import time

a = np.random.rand(1000000)
b = np.random.rand(1000000)

tic = time.time()
c = dot(a,b)
toc = time.time()
print("Dot product: ", c)
print ("Time for the loop version:" + str(1000*(toc-tic)) + " ms")

tic = time.time()
c = np.dot(a,b)
toc = time.time()
print("Dot product: ", c)
print ("Time for the vectorized version, np.dot() function: " + str(1000*(toc-tic)) + " ms")

tic = time.time()
c = a @ b
toc = time.time()
print("Dot product: ", c)
print ("Time for the vectorized version, @ function: " + str(1000*(toc-tic)) + " ms")
```

```python
# Dot Product of orthogonal vectors are zeros
i = np.array([1, 0, 0])
j = np.array([0, 1, 0])
print("The dot product of i and j is", dot(i, j))
```
