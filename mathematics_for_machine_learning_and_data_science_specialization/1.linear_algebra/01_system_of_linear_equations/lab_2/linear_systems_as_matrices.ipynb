# Representing Systems of Equations as Matrices

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
arr1 = np.array([[-1,3], [3,2]], dtype=np.dtype(float))
arr2 = np.array([7,1], dtype=np.dtype(float))

print(arr1)
print(arr2)

print(f"Shape of A: {arr1.shape}")
print(f"Shape of b: {arr2.shape}")
```

```python
x = np.linalg.solve(arr1, arr2)
print(x)
print(f"Solution: {x}")
```
