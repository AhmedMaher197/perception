### **Errors and Exceptions**

```python
# --- Syntax Errors ---
# occur when Python can’t interpret our code, since we didn’t follow the correct syntax for Python.

# --- Exceptions ---
# occur when unexpected things happen during execution of a program, 
# even if the code is syntactically correct. 
# There are different types of built-in exceptions in Python, 
# and you can see which exception is thrown in the error message.
```

```python
# --- ValueError Exception ---
# An object of the correct type but inappropriate value is passed as input to built-in 
# operation or function.
number = int("hello")
```

```python
# --- AssertionError Exception ---
# AssertionError occurs when an assert statement fails
x = 5
assert x == 10
```

```python
# --- IndexError Exception ---
# try to access element which out of bound in an array
numbers = [1, 2, 3]
print(numbers[5])  # IndexError: list index out of range
```

```python
# --- KeyError Exception ---
# A key can't be found in a dictionary.

student = {'name': 'John', 'age': 20}
print(student['grade'])  # KeyError: 'grade'
```

```python
# --- TypeError Exception ---
# An object of an unsupported type is passed as input to an operation or function.
# Mixing incompatible types:
result = "5" + 3  # TypeError: can only concatenate str (not "int") to str  
```

### **try and except**

```python
# --- Simple try except ---
try:
    x = int('Enter a number: ')
except:
    print("Input not a valid number")
```

```python
# Looping with try except 
c = 0
while True:
    try:
        if c == 0:
            c = c + 1
            # wrong input assigned to x
            x = int('string')
        else:
            # correct input assigned to x
            x = 5
        break
    except:
        print("Input not a valid number")
    
    print("While loop ended.")
```

```python
# try except else finally execution sequence
c = 0
# else key word in try except
while c != 1:
    try:
        print("In try")
        if c == 0:
            c = 1
            x = 5
            # by executing this line except shall be called and else section will not
            # be executed
            # x = int('Hello')
        else:
            x = 1
        # Will break while loop + try..except.
    except:
        print("Input not a valid number")
    # else section is only run if not exception detected.
    else:
        print("In else")
        y = 2
        c = 3
        # will throw ValueError Exception
        # c = int('hello')
        break
    # execute wether there is any exception occur or not
    finally:
        print("Input Attempt")
```

```python
# --- Handle specificed exceptions ---
c = 0
while c != 2:
    try:
        c = c + 1
        x = int("Hello")
        break
    except ValueError:
        print('Not a valid number!')
    finally:
        print("Attempted Input")
```

```python
# --- Handle multiple exceptions ---
c = 0
while c != 2:
    try:
        print("In try")
        if c == 0:
            c += 1
            x = int("Hello")
        if c == 1:
            c += 1
            x = "5" + 3
        break
    except (ValueError, TypeError):
        print('Not a valid number!')
    finally:
        print('Attempted Input')
```

```python
# --- Handle multiple exceptions ---
c = 0
while c != 2:
    try:
        print("In try")
        if c == 0:
            c += 1
            x = int("Hello")
        if c == 1:
            c += 1
            x = "5" + 3
        break
    except ValueError:
         print('ValueError exception')
    except TypeError:
         print('TypeError exception')
    finally:
        print('Attempted Input')
```

```python
# Accessing Error Messages
try:
    # some code
    print (10 / 0)
except ZeroDivisionError as e:
   # some code
   print("ZeroDivisionError occurred: {}".format(e))
```

### **Importing Local Scripts**

```python
# --- Importing Local Scripts ---
import useful_functions
print(type(useful_functions))
# access var in the other script
print(useful_functions.num_useful_func)

# use funtions from the other script
scores = [88, 92, 79, 93, 85]
mean = useful_functions.mean(scores)
curved = useful_functions.add_five(scores)
mean_c = useful_functions.mean(curved)
print("Scores:", scores)
print("Original Mean:", mean, "  New Mean:", mean_c)

# alisa for script name with as keyword
import useful_functions as uf
print(uf.num_useful_func)
```
