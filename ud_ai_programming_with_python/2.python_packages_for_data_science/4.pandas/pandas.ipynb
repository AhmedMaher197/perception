# **Pandas**

### **Category: Initialization and Utility**

| Function/Method | Description |
|-----------------|-------------|
| `pandas.read_csv(relative_path_to_file)` | Reads a comma-separated values (csv) file present at relative_path_to_file and loads it as a DataFrame |
| `pandas.DataFrame(data)` | Returns a 2-D heterogeneous tabular data. Note: There are other optional arguments as well that you can use to create a dataframe |
| `pandas.Series(data, index)` | Returns 1-D ndarray with axis labels |
| `pandas.Series.shape` / `pandas.DataFrame.shape` | Returns a tuple representing the dimensions |
| `pandas.Series.ndim` / `pandas.DataFrame.ndim` | Returns the number of dimensions (rank). It will return 1 in case of a Series |
| `pandas.Series.size` / `pandas.DataFrame.size` | Returns the number of elements |
| `pandas.Series.values` | Returns the data available in the Series |
| `pandas.Series.index` | Returns the indexes available in the Series |
| `pandas.DataFrame.isnull()` | Returns a same sized object having True for NaN elements and False otherwise |
| `pandas.DataFrame.count(axis)` | Returns the count of non-NaN values along the given axis. If axis=0, it will count down the dataframe, meaning column-wise count of non-NaN values |
| `pandas.DataFrame.head([n])` | Return the first n rows from the dataframe. By default, n=5 |
| `pandas.DataFrame.tail([n])` | Return the last n rows from the dataframe. By default, n=5. Supports negative indexing as well |
| `pandas.DataFrame.describe()` | Generate the descriptive statistics, such as, count, mean, std deviation, min, and max |
| `pandas.DataFrame.min()` | Returns the minimum of the values along the given axis |
| `pandas.DataFrame.max()` | Returns the maximum of the values along the given axis |
| `pandas.DataFrame.mean()` | Returns the mean of the values along the given axis |
| `pandas.DataFrame.corr()` | Compute pairwise correlation of columns, excluding NA/null values |
| `pandas.DataFrame.rolling(window)` | Provide rolling window calculations, such as pandas.DataFrame.rolling(15).mean() for rolling mean over window size of 15 |
| `pandas.DataFrame.loc[label]` | Access a group of rows and columns by label(s) |
| `pandas.DataFrame.groupby(mapping_function)` | Groups the dataframe using a given mapper function or by a Series of columns |

### **Category: Manipulation**

| Function/Method | Description |
|-----------------|-------------|
| `pandas.Series.drop(index)` | Drops the element positioned at the given index(es) |
| `pandas.DataFrame.drop(labels)` | Drop specified labels (entire columns or rows) from the dataframe |
| `pandas.DataFrame.pop(item)` | Return the item and drop it from the frame. If not found, then raise a KeyError |
| `pandas.DataFrame.insert(location, column, values)` | Insert column having given values into DataFrame at specified location |
| `pandas.DataFrame.rename(dictionary-like)` | Rename label(s) (columns or row-indexes) as mentioned in the dictionary-like |
| `pandas.DataFrame.set_index(keys)` | Set the DataFrame's row-indexes using existing column-values |
| `pandas.DataFrame.dropna(axis)` | Remove rows (if axis=0) or columns (if axis=1) that contain missing values |
| `pandas.DataFrame.fillna(value, method, axis)` | Replace NaN values with the specified value along the given axis, and using the given method ('backfill', 'bfill', 'pad', 'ffill', None) |
| `pandas.DataFrame.interpolate(method, axis)` | Replace the NaN values with the estimated value calculated using the given method along the given axis |

### **Pandas Series**

```python
import numpy as np
import pandas as pd
```

```python
# We create a Pandas Series that stores a grocery list
groceries = pd.Series(data = [30, 6, 'Yes', 'No'], index = ['eggs', 'apples', 'milk', 'bread'])

# We display the Groceries Pandas Series
print("groceries details:")
print(groceries)

# shape to return data(rows, cols)
print("groc shape:", groceries.shape)

# ndim to show rand of the data
print("groc rank:", groceries.ndim)

# to show data size
print("groc size:", groceries.size)

# To check index labels of a series
print("groc indexes:", groceries.index)

# To check values of data in a series
print("groc values:", groceries.values)

# Check if data exist in series
print("Is Banana exist in groc:", 'banana' in groceries)
print("Is bread exist in groc:", 'bread' in groceries)
```

### **Accessing and Deleting Elements in pandas Series**

```python
groceries = pd.Series(data = [30, 6, 'Yes', 'No'], index = ['eggs', 'apples', 'milk', 'bread'])
print("groceries details:")
print(groceries)
```

```python
# filter on a value with key
print("groceries['eggs'] =", groceries['eggs'])

# filter on multiple values with multiple keys
print("groceries[['milk', 'bread']] =")
print(groceries[['milk', 'bread']])

# use normal indexing to get the first element
print("groceries[0] =", groceries[0])

# use normal indexing to get the last element
print("groceries[-1] =", groceries[-1])

# we can provide list of indices to get their values
print("groceries[[0,1]]:")
print(groceries[[0,1]])
```

```python
# .loc Label-based indexing to get the values of an index
print("groceries.loc[['eggs', 'apples']] =")
print(groceries.loc[['eggs', 'apples']])

# .iloc Integer position-based indexing to get the values of an index
print("groceries.iloc[[2, 3]] =")
print(groceries.iloc[[2, 3]])
```

```python
# Panda Series are mutable same as NumPy Array which means we can edit in the series after creating it
groceries['eggs'] = 2
print("groceries after modifying eggs:")
print(groceries)
```

```python
# .drop proivde a copy of the Panda series after removing an element, original series still the same
print("Copy Series of Gros after removing apples:")
print(groceries.drop('apples'))

# apples still not removed
print("original series:")
print(groceries)

# to delete an element in the original series.
groceries.drop('apples', inplace=True)
print("original series after reomve apples:")
print(groceries)
```

### **Arithmetic Operations on Pandas Series**

```python
fruits = pd.Series([10,6,3], ['apples', 'oranges', 'banans'])
print(fruits)
```

```python
# Apply Arthmitic operations on all panda series
# +
print(fruits + 2)
# -
print(fruits - 2)
# *
print(fruits * 2)
# / 
print(fruits / 2)
```

```python
# we can use numpy use statical operation on pandas list
import numpy as np

# sqrt
print(np.sqrt(fruits))

# exp
print(np.exp(fruits))

# power
print(np.power(fruits, 2))
```

```python
# we can apply arithmetic operations on pandas elements
print("updated banans = ", fruits['banans'] + 2)
print("updated apples = ", fruits.iloc[0] - 2)

print("double apples, oranges")
print(fruits[['banans', 'apples']] * 2)

print("devide apples, oranges by 2")
print(fruits[['banans', 'oranges']] / 2)
```

```python
# Apply arithmetic operations on series which include different data types
groceries = pd.Series(data = [30, 6, 'Yes', 'No'], index = ['eggs', 'apples', 'milk', 'bread'])
print("groceries details:")
print(groceries)

# Multiplication operation is defined for both strings and numbers so no errors
print("Multiply the series by 2:")
print(groceries * 2)

# if we try division which can't be done with string then an error will be thrown
# print(groceries / 2) 
```

### **Creating Pandas DataFrames**

```python
# Data Frame is a 2 dimensional object with labeled rows and cols,
# and can also hold multiple data types

# Create a dictionnary of 2 Pandas series that represent itmes bought by 2 persons.
items = {'Alice': pd.Series(data = [40, 110, 500, 45], index = ['book', 'glasses', 'bike', 'pants']),
         'Bob': pd.Series(data = [245, 25, 55], index = ['bike', 'pants', 'watch'])}

print(type(items))
```

```python
# Using DataFrame
shopping_carts = pd.DataFrame(items)

# DataFrame view as a tabuler form same as spreadsheet
# Labels of rows and cols are in bold
# cols labels are taken from the keys of the dictionary
# row labels are taken from the union of the indexes of panda series
# NaN stands for "Not a Number"
shopping_carts
```

```python
# we can extract information from panda series using attributes
print(shopping_carts.index)
print(shopping_carts.columns)
print(shopping_carts.values)

# we can also pringf the size of the DataFrame
print(shopping_carts.shape)

# also we can print the rank of the frame
print(shopping_carts.ndim)

# can get the size if the DataFrame
print(shopping_carts.size)
```

```python
# we might be intersted in a part or section from series to be a data frame
bob_shopping_cart = pd.DataFrame(items, columns = ['Bob'])
print("Bob items:")
print(bob_shopping_cart)

# we can also be intersted in a specific items
sel_shopping_cart = pd.DataFrame(items, index = ['pants', 'book'])
print("pants, book items in shopping_carts:")
print(sel_shopping_cart)

# and we might be intersted in a specific items for a specific person
alice_sel_shopping_cart = pd.DataFrame(items, index=['glasses', 'bike'], columns=['Alice'])
print("pants, book items in Alice:")
print(alice_sel_shopping_cart)
```

### **Accessing Elements in Pandas DataFrames**

```python
items = [{'bikes': 20, 'pants': 30, 'watches': 35}, 
          {'watches': 10, 'glasses': 50, 'bikes': 15, 'pants':5}]

store_items = pd.DataFrame(items, index=['store1', 'store2'])
store_items
```

```python
# By using rows and cols labels we can access elements in DataFrame
# cols indexing
print("Access bikes col labels:")
print(store_items[['bikes']])

# access multiple columns
print("Access multiple cols labels:")
print(store_items[['bikes', 'pants']])

# row indexing
print("store1 data:")
print(store_items.loc['store1'])

# access specific element in a row
# note that cols indexing comes first not the same as normal arrays
print("store_items['bikes']['store2'] = ", store_items['bikes']['store2'])

# error, cols indexing must come first
# print("store_items['store2']['bikes'] = ", store_items['store2']['bikes'])
```

```python
# Add New cols to dataFrame
# Use syntax in line 5 will lead to error because the cols needs to be defined first for both stores then the change in the valus come
# store_items['shirts']['store2'] = 14

# To add new index with data into data frame
store_items['shirts'] = np.nan
store_items['shirts']['store1'] = 20
store_items['shirts']['store2'] = 15
print(store_items)

# add and mofify values of a col
store_items['blouses'] = 15
print(store_items)

# add suits to both stores respecting the values for each formula
store_items['suits'] = store_items['shirts'] + store_items['pants']
print(store_items)
```

```python
# Add new row to existed DataFrame
# 1. Defined the new store to be a new data frame
new_items = [{'bikes': 20, 'pants': 30, 'watches': 35, 'glasses': 4}]
new_store = pd.DataFrame(new_items, index = ['store3'])
print(new_store)

# 2. appned the new data frame to the one which we need to update it
store_items = pd.concat([store_items, new_store])
print(store_items)
```

```python
# Add new cols diff indexing
# skip adding wacthed of store1 to the new col "new_Watches"
store_items['new_watches'] = store_items['watches'][1:]
store_items
```

```python
# insert method to add new cols to DataFrame
# insert(index, 'new_cols_name', [data0,1,2])
store_items.insert(5, 'shoes', [8,5,0])
print(store_items)

store_items.insert(5, 'slippers', 8)
print(store_items)
```

```python
# pop('col_name') used to remove cols from DataFrame
store_items.pop('new_watches')
print(store_items)
```

```python
# drop method to remove cols and rows from DataFrame
# drop([cols,...], axis=1), return a copy of the origing DataFrame after removing cols.
store_items = store_items.drop(['watches', 'shoes'], axis=1)
print(store_items)

# drop([rows,...], axis=0), return a copy of the origing DataFrame after removing cols.
store_items = store_items.drop(['store1', 'store2'], axis=0)
print(store_items)
```

```python
# rename method to rename labels in cols and rows
# rename col index
store_items = store_items.rename(columns={'bikes' : 'hats'})

# rename row index
store_items = store_items.rename(index={'store3' : 'last_store'})
print(store_items)
```

```python
# set_index('index'), converts a column into the row index of the DataFrame.
store_items = store_items.set_index('pants')
print(store_items)

# accessing rows using pants elementes
print(store_items.loc[30])
print(store_items.loc[5])
```

### **Dealing with NaN**

```python
# NaN values consider to be a bad data that we need to filter it
items = [{'bikes': 20, 'pants': 30, 'watches': 35, 'shirts': 15, 'shoes': 8, 'suits': 45},
         {'watches': 10, 'glasses': 50, 'bikes': 15, 'pants': 5, 'shirts': 2, 'shoes': 5, 'suits': 7},
         {'bikes': 20, 'pants': 30, 'watches': 35, 'glasses': 4, 'shoes': 10}]

store_items = pd.DataFrame(items, index=['store 1', 'store 2', 'store 3'])
store_items
```

```python
# Count numbers of NaN values
# printing the Frame with indications to the values that equals to Null "True"
x = store_items.isnull()

# provide total number of NaN values in each col of the DataFrame
x = store_items.isnull().sum()

# provide the total number of NaN values in the DataFrame
x = store_items.isnull().sum().sum()

print("Total number of NaN values: ",x)
```

```python
# We can also count number of non NaN values in each col
print(store_items.count())

# count number of non NaN values in DataFrame
print("Total number of non NaN values: ", store_items.count().sum())
print("Total number of NaN values: ", store_items.size - store_items.count().sum())
```

```python
# dropna(axis=0/1), return a new DataFrame after rmoving any row / cols which contain NaN values
print("Original Data Frame:")
print(store_items)

print("Data Frame after removing NaN values from rows:")
print(store_items.dropna(axis=0))

print("Data Frame after removing NaN values from cols:")
print(store_items.dropna(axis=1))

# To Apply the modifications in the original DataFrame
# store_items.dropna(axis=1, inplace=True)
# store_items.dropna(axis=0, inplace=True)

# original list still the same
print("Original Data Frame stay the same:")
print(store_items)
```

```python
print("Original DF:")
print(store_items)

# fillna(value) used to return new DataFrame that fill the NaN values in rows and cols with values.
print("filling NaN values in DF with zeros:")
print(store_items.fillna(4))

# ffill(axis=0/1) forward filling, NaN values will be filled with the values that being exist in the previous axis.
print("filling NaN values in DF with prvious axis values rows:")
print(store_items.ffill(axis=0))

print("filling NaN values in DF with prvious axis values col:")
print(store_items.ffill(axis=1))

# To Apply the modifications in the original DataFrame
# store_items.ffill(axis=1, inplace=True)
# store_items.ffill(axis=0, inplace=True)
```

```python
print("Original DF:")
print(store_items)

# bfill(axis=0/1) backward filling, NaN values will be filled with the values that being exist in the next axis.
print("filling NaN values in DF with next axis values rows:")
print(store_items.bfill(axis=0))

print("filling NaN values in DF with next axis values col:")
print(store_items.bfill(axis=1))

# To Apply the modifications in the original DataFrame
# store_items.bfill(axis=1, inplace=True)
# store_items.bfill(axis=0, inplace=True)
```

```python
print("Original DF:")
print(store_items)

print("\nAfter interpolation:")
# Search about what inerploation means
print(store_items.interpolate(method='linear', axis=0))
print(store_items.interpolate(method='linear', axis=1))
```
