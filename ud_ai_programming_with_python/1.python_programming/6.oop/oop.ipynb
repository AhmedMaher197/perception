### **args and kwargs Keywords**

```python
# *args
# The * symbol tells Python "collect all the positional arguments into a tuple called args.
# You can name it anything you want, but args is the conventional name.

def greet_people(*args):
    """This function can greet any number of people"""
    print(f"Arguments received: {args}")  # args is a tuple
    print(f"Type of args: {type(args)}")
    
    for person in args:
        print(f"Hello {person}!")

# Test with different numbers of arguments
print("Testing with 1 argument:")
greet_people("Alice")                    # 1 argument

print("\nTesting with 2 arguments:")
greet_people("Alice", "Bob")             # 2 arguments  

print("\nTesting with 3 arguments:")
greet_people("Alice", "Bob", "Charlie")  # 3 arguments
```

```python
# **kwargs
# Dictionary the collect all the arguments with their defaults
def describe_person(**kwargs):
    """This function can accept any number of keyword arguments"""
    print(f"Arguments received: {kwargs}")  # kwargs is a dictionary
    print(f"Type of kwargs: {type(kwargs)}")
    
    for key, value in kwargs.items():
        print(f"{key}: {value}")

# Test with different keyword arguments
print("Testing with basic info:")
describe_person(name="Alice", age=30)

print("\nTesting with extended info:")
describe_person(name="Bob", age=25, city="New York", job="Engineer")
```

```python
# Combining Both: The Ultimate Flexibility
def flexible_function(*args, **kwargs):
    """This function accepts any arguments you throw at it"""
    print("Positional arguments:")
    for i, arg in enumerate(args):
        print(f"  {i}: {arg}")
    
    print("Keyword arguments:")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

# Test with various combinations
print("Testing with mixed arguments:")
flexible_function(1, 2, 3, name="Alice", age=30)

print("Testing with different combination:")
flexible_function("hello", greeting="Hi", count=5)

print("Testing with no arguments:")
flexible_function()  # Even no arguments works!
```

```python
# UnPacking usin * and **
def add_three_numbers(a, b, c):
    return a + b + c

# Instead of calling it like this:
numbers = [1, 2, 3]
result = add_three_numbers(numbers[0], numbers[1], numbers[2])
print(f"Manual unpacking result: {result}")

# You can unpack the list with *:
numbers = [1, 2, 3]
# Python expands this to add_three_numbers(1, 2, 3)
result = add_three_numbers(*numbers)  
print(f"Automatic unpacking result: {result}")  # Output: 6
```

```python
def introduce_person(name, age, city):
    print(f"Hi, I'm {name}, {age} years old, from {city}")

# Instead of this:
person_info = {"name": "Alice", "age": 30, "city": "Boston"}
print("Manual unpacking:")
introduce_person(person_info["name"], person_info["age"], person_info["city"])

# You can unpack the dictionary with **:
person_info = {"name": "Alice", "age": 30, "city": "Boston"}
print("\nAutomatic unpacking:")
# Python expands this to introduce_person(name="Alice", age=30, city="Boston")
introduce_person(**person_info)
```

### **Intro to OOP**

```python
# When you first learn to program, you are probably using a style called procedural programming.
# A procedural program is essentially a list of instructions that execute one after the other,
# starting from the top of a file and working down.

# Object-oriented programs, on the other hand, are built around well... objects!
# You can think about objects as things that exist in the real world.

# PROCEDURAL PROGRAMMING:
# - A list of instructions that execute sequentially
# - Top-to-bottom execution
# - Focus on functions and procedures
# - Data and functions are separate

# OBJECT-ORIENTED PROGRAMMING (OOP):
# - Built around objects that model real-world things
# - Focus on objects that have characteristics and actions
# - Data and methods are bundled together
# - More intuitive way to model complex systems

# If you were going to build a clothing store application, think about all the objects:
# - The store itself would be an object
# - The items in the store like shirts and jeans would also be objects  
# - The cash register would be an object
# - Even a salesperson would be an object

# An object-oriented program would focus on the individual characteristics 
# of each object and what each object can do.

# An object has TWO essential parts:
# 1. CHARACTERISTICS (what it IS) (also called attributes or properties)
# 2. ACTIONS (what it can DO) (also called methods or behaviors)

# Example 1: Salesperson Object
# CHARACTERISTICS of a salesperson:
# - Name
# - Address  
# - Phone number
# - Hourly pay

# ACTIONS a salesperson can do:
# - Sell an item
# - Take an item from warehouse
# - Help customers
# - Process returns

# ============================================================= #

# Example 2: Shirt Object
# CHARACTERISTICS of a shirt:
# - Color
# - Size
# - Style  
# - Price

# ACTIONS a shirt can take:
# It might sound silly to ask what actions a shirt could take since a shirt 
# is an inanimate object, but a shirt could, for example:
# - Change its price
# - Be added to cart
# - Be removed from inventory
# - Update its availability
```

```python
# --- Essential OOP Terms ---
# CLASS
# - A blueprint consisting of methods and attributes
# - Think of it as a template or recipe for creating objects
# - Defines what characteristics and actions objects will have

# OBJECT  
# - An instance of a class
# - It can help to think of objects as something in the real world like:
#   * a yellow pencil
#   * a small dog  
#   * a blue shirt
# - However, as you'll see later in the lesson, objects can be more abstract

# ATTRIBUTE
# - A descriptor or characteristic  
# - Examples would be color, length, size, etc.
# - These attributes can take on specific values like blue, 3 inches, large, etc.

# METHOD
# - An action that a class or object could take
# - These are functions that belong to the class/object

# OOP
# - A commonly used abbreviation for object-oriented programming

# ENCAPSULATION
# - One of the fundamental ideas behind object-oriented programming
# - You can combine functions and data all into a single entity
# - In object-oriented programming, this single entity is called a class
# - Encapsulation allows you to hide implementation details much like how 
#   the scikit-learn package hides the implementation of machine learning algorithms
```

```python
# --- Creating a Shirt Class ---
class Shirt:
    # The __init__ method is special - it's called when you create a new object
    # This is known as the "constructor" or "initializer"
    # 'self' refers to the specific instance being created
    def __init__(self, shirt_color, shirt_size, shirt_style, shirt_price):
        # These are the ATTRIBUTES (characteristics) of our shirt
        # We assign the parameter values to instance attributes using 'self' -> blueprint
        self.color = shirt_color
        self.size = shirt_size
        self.style = shirt_style
        self.price = shirt_price
    
    # This is a METHOD (action) that the shirt can perform
    def change_price(self, new_price):
        # This method allows us to update the shirt's price
        self.price = new_price
    
    # Another METHOD - calculate discounted price
    def discount(self, discount):
        # This method returns the discounted price without changing the original price
        return self.price * (1 - discount)


new_shirt = Shirt('red', 'S', 'short sleeve', 15)
print(type(new_shirt))

# Now let's access the attributes of our shirt object
# a) "--- Accessing Object Attributes ---"
print("Shirt color:", new_shirt.color)
print("Shirt size:", new_shirt.size)
print("Shirt style:", new_shirt.style)
print("Shirt price:", new_shirt.price)

# b) "--- Accessing Object Methods ---"
new_shirt.change_price(10)
print("Shirt price:", new_shirt.price)

print("Shirt price after discount:", new_shirt.discount(.2))

# List with Objects
print("\n--- Working with Lists of Objects ---")

# Create a list to hold our shirt collection
tshirt_collection = []

# Create individual shirt objects
shirt_one = Shirt('orange', 'M', 'short sleeve', 25)
shirt_two = Shirt('red', 'S', 'short sleeve', 15)
shirt_three = Shirt('purple', 'XL', 'short sleeve', 10)

# Add the shirt objects to our collection
tshirt_collection.append(shirt_one)
tshirt_collection.append(shirt_two)
tshirt_collection.append(shirt_three)

print("Created a collection of", len(tshirt_collection), "shirts")

# Loop through the collection and access object attributes
print("\nShirt colors in our collection:")
for i in range(len(tshirt_collection)):
    print(f"Shirt {i+1} color:", tshirt_collection[i].color)

# Note Python is not like the C++ to have private and public methods and attributes
# but still it is recommended to use settes and getters for the private members of the class
# private member in python are pre-fixed with _ as an indication that it is private member
# and shall not accessed directly.
```

### **OOP Class Magic methods**

```python
# Magic methods in Python are special methods that start and
# end with double underscores (__). They allow you to define
# how objects of your class behave with built-in Python
# operations.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    # __repr__ method - returns a string representation for developers
    # This should be unambiguous and ideally eval(repr(obj)) == obj)
    def __repr__(self):
        return f"Point({self.x}, {self.y})"
    
    # __str__ method - returns a human-readable string representation
    # This is used by print() function
    def __str__(self):
        return f"({self.x}, {self.y})"
    
    # __add__ method - defines behavior for + operator
    def __add__(self, other):
        if isinstance(other, Point):
            return Point(self.x + other.x, self.y + other.y)
        return NotImplemented
    
    # __eq__ method - defines behavior for == operator
    def __eq__(self, other):
        if isinstance(other, Point):
            return self.x == other.x and self.y == other.y
        return False

# Create Point objects
p1 = Point(2, 3)
p2 = Point(4, 5)

print("\n--- Testing Magic Methods ---")

# Test __str__ (used by print)
print("Point 1:", p1)  # Uses __str__
print("Point 2:", p2)  # Uses __str__

# Test __repr__ (used in interactive shell and debugging)
print("repr(p1):", repr(p1))  # Uses __repr__
print("repr(p2):", repr(p2))  # Uses __repr__

# Test __add__ (addition operator)
p3 = p1 + p2  # Uses __add__
print("p1 + p2 =", p3)

# Test __eq__ (equality operator)
p4 = Point(2, 3)
print("p1 == p4:", p1 == p4)  # Uses __eq__ - should be True
print("p1 == p2:", p1 == p2)  # Uses __eq__ - should be False
```

### **Inheritance**

```python
# Inheritance is one of the core principles of Object-Oriented Programming.
# It allows a class (child/derived class) to inherit attributes and methods 
# from another class (parent/base class).
# PARENT CLASS (Base Class)
# This is the general class that contains common attributes and methods
class Clothing:
    """Parent class for all clothing items"""
    
    def __init__(self, color, size, style, price):
        # These attributes are shared by ALL clothing items
        self.color = color
        self.size = size
        self.style = style
        self.price = price
    
    def change_price(self, new_price):
        """Method to change the price - available to all clothing items"""
        self.price = new_price
    
    def calculate_discount(self, discount):
        """Method to calculate discounted price - available to all clothing items"""
        return self.price * (1 - discount)
    
# --- Child Classes (Inheriting from Clothing) ---
# CHILD CLASS 1 (Derived Class)
# Shirt inherits from Clothing and adds its own specific attributes/methods
class Shirt(Clothing):
    """Child class for shirts - inherits from Clothing"""
    
    def __init__(self, color, size, style, price, long_or_short):
        # Call the parent class constructor using super()
        # This initializes color, size, style, price from the parent class
        Clothing.__init__(self, color, size, style, price)
        # Alternative syntax: super().__init__(color, size, style, price)
        
        # Add shirt-specific attribute
        self.long_or_short = long_or_short
    
    def double_price(self):
        """Shirt-specific method - only shirts can double their price"""
        self.price = 2 * self.price

# CHILD CLASS 2 (Derived Class)  
# Pants inherits from Clothing and adds its own specific attributes/methods
class Pants(Clothing):
    """Child class for pants - inherits from Clothing"""
    
    def __init__(self, color, size, style, price, waist):
        # Call the parent class constructor
        Clothing.__init__(self, color, size, style, price)
        
        # Add pants-specific attribute
        self.waist = waist
    
    def calculate_discount(self, discount):
        """Override parent method - pants get half the discount"""
        return self.price * (1 - discount / 2)
    
# Create instances of child classes
shirt = Shirt('blue', 'M', 'long sleeve', 20, 'long')
pants = Pants('black', 'L', 'skinny', 30, 32)

print("=== Shirt Object ===")
print(f"Color: {shirt.color}")           # Inherited from Clothing
print(f"Size: {shirt.size}")             # Inherited from Clothing  
print(f"Style: {shirt.style}")           # Inherited from Clothing
print(f"Price: ${shirt.price}")          # Inherited from Clothing
print(f"Sleeve: {shirt.long_or_short}")  # Shirt-specific attribute

print(f"Discount price: ${shirt.calculate_discount(0.2)}")  # Inherited method
shirt.double_price()                     # Shirt-specific method
print(f"After doubling: ${shirt.price}")

print("\n=== Pants Object ===")
print(f"Color: {pants.color}")           # Inherited from Clothing
print(f"Size: {pants.size}")             # Inherited from Clothing
print(f"Style: {pants.style}")           # Inherited from Clothing  
print(f"Price: ${pants.price}")          # Inherited from Clothing
print(f"Waist: {pants.waist}")           # Pants-specific attribute

print(f"Discount price: ${pants.calculate_discount(0.2)}")  # Overridden method
```
