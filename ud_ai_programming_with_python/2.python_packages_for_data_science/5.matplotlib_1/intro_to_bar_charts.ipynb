# **Intro to Bar Charts**

### **Tidy dataset is a tabular dataset where:**

- Each variable is a column
- Each observation is a row
- Each type of observational unit is a table



### **Bar Chart**
 - Used with qualitative variables
 - Each level ofthe categorical variable has a unique x position 
 - The height of each bar illustrates the frequency or count of the number of times a data point took on each categorical value.



### **Example of using Bar Chart:**

**Here is it visiable that:**

 - Male bar height: 450 
 - Female bar height: 580 
 - Visual ratio = 580/450 = 1.29

**But in case we start from 400**

 - Male bar height: 450 - 400 = 50 units
 - Female bar height: 580 - 400 = 180 units
 - Visual ratio: 180/50 = 3.6 (appears 3.6 times larger)



### **Nominal data:**

 - Arrange the bars in order of frequency with the most common on the left.

 - This makes it easy to see which category is the most common and how the rest of the categories compare



### **Ordianl Data**

 - The inherent order of the levels will usually be a more important feature to convey.

 - It's more important to know if the most common categories are loweror higher in the spectrum rather than just which label is the most frequent.



### **Ordinal data vs nominal data**

#### <u>**Ordinal data with inherent order:**</u>

<u>Ex1:</u>

 - Education levels: Elementary → High School → College → Graduate

 - This order exists naturally - you can't rearrange it randomly

<u>Ex2:</u>

 - Survey ratings: Strongly Disagree → Disagree → Neutral → Agree → Strongly Agree

 - The progression from negative to positive is built into the meaning



#### <u>**Compare to nominal data (no inherent order):**</u>

<u>Ex1:</u>

 - Colors: Red, Blue, Green, Yellow

 - No natural order - you could arrange them any way

<u>Ex2:</u>

 - Countries: USA, Canada, Mexico, Brazil

 - No inherent ranking - any order is arbitrary

### **Horizontal Bar Chart:**

- The chart has essentially turned 90 degrees clockwisesuch that the Y positions indicate category and the X-linked indicates frequency.
- If you have a lot of categories or the category names are long,then the horizontal orientation can be moreconvenient in terms of the space needed to create a readable plot.



### **Absolute Vs Relative frequency**


<u>Relative Frequency</u>

The proportion of the data that falls in each category

<u>Ex1:</u>

To get relative frequency, you divide each count by the total:
- Total days = 744 + 252 + 99 = 1,095 days
- Good: 744 ÷ 1,095 = 0.68
- Moderate: 252 ÷ 1,095 = 0.23
- Unhealthy: 99 ÷ 1,095 = 0.09
