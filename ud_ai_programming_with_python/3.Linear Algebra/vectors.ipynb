```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a vector
vector = np.array([1,2])

# plot vector
plt.quiver(0, 0, vector[0], vector[1], angles='xy', scale_units='xy', scale=1)
plt.xlim(0, 4)
plt.ylim(0, 3)

plt.grid()
plt.show()
```

```python
# Vector Scaling
scaled_vector = 2 * vector

# plot vector
# Note try to draw the smallest first
plt.quiver(0, 0, scaled_vector[0], scaled_vector[1], angles='xy', scale_units='xy', scale=1, color='green')
plt.quiver(0, 0, vector[0], vector[1], angles='xy', scale_units='xy', scale=1, color='red')
plt.xlim(0, 4)
plt.ylim(0, 5)

plt.grid()
plt.show()
```

```python
# Vector Addition
v1 = np.array([1,2])
v2 = np.array([3,-1])
v3 = v1 + v2

plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='green')
plt.quiver(v1[0], v1[1], v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='blue')
plt.quiver(0, 0, v3[0], v3[1], angles='xy', scale_units='xy', scale=1, color='red')

plt.xlim(0, 4)
plt.ylim(-3, 5)
plt.grid()
plt.show()
```

```python
# Note
# With a 1D NumPy array like np.array([3, 2]), the transpose won't change anything because it's neither a row nor column vector - it's just a 1D array.

# Vector Transpose
v = np.array([[3,5]])
v = v.T
print (v)
```
