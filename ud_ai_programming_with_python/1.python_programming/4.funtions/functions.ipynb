# **Funtions**

```python
# Function with default value
def cylinder_volume(height, radius=5):
    pi = 3.14159
    return height * pi * radius ** 2

print("cylinder_volume(10) = ", cylinder_volume(10))
print("cylinder_volume(10,5) = ", cylinder_volume(10, 5))
```

```python
# Pass arguments to functions
def cylinder_volume(height, radius=5):
    pi = 3.14159
    return height * pi * radius ** 2

print("By Value:")
print("cylinder_volume(10, 7)", cylinder_volume(10, 7))

print("By Position:")
print("cylinder_volume(height=10, radius=2)", cylinder_volume(height=10, radius=2))
print("Order does not matter")
print("cylinder_volume(radius=2, height=10)", cylinder_volume(radius=2, height=10))
```

### **Global Variable with Function**

```python
# use global keyword to reflect the value on the global variables
egg_count = 10
def buy_eggs():
    # "global" must be used in order to make egg_count override.
    global egg_count
    print("Sol1: global egg_count = ", egg_count)
    egg_count += 12 # purchase a dozen eggs
buy_eggs()
print("global egg_count = ", egg_count)
```

```python
# OR overwrite the global variable out of function scope
egg_count = 10
def buy_eggs(count):
    return count + 12  # purchase a dozen eggs
egg_count = buy_eggs(egg_count)
print("Sol2: egg_count = ", egg_count)
```

```python
print("4.Same var name can be used in different scopes")
str1 = 'Functions are important programming concepts.'
def print_fn():
    # Only visible in print_fn
    str1 = 'Variable scope is an important concept.'
    print("print_fn scope:")
    print(str1)

print_fn()
print("Main scope:")
print(str1)
```

### **Lambda Functions**

```python
def double_default(x):
    return x * 2
print("double_default(2) = ", double_default(2))

double_lambda = lambda x: x *2
print("double_lambda(2) = ", double_lambda(2))
```

```python
def multiply_default (x, y):
    return x * y
print("multiply_default(2, 3) = ", multiply_default(2, 3))

multiply_lambda = lambda x, y: x * y
print("multiply_lambda(2, 3) = ", multiply_lambda(2, 3))
```
