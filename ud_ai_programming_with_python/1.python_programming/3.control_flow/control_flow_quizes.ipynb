### **Quiz 21**

```python
# Scenario:
# ==========
# You are training a machine learning model, and you want to stop the
# training process if the validation loss does not improve for a certain
# number of epochs. You have a list of tuples where each tuple represents
# an epoch's performance, with the first element being the epoch number and
# the second element being the validation loss.
# ==========
# Task:
# ==========
# Method 1: Stop the training immediately when the validation loss increases
# compared to the previous epoch. Method 2: Stop the training when the
# validation loss has not improved for a certain number of epochs (patience).
# ==========
# Solution:
# ==========
print("Quiz 21: Early Stopping Methods")

epoch_performances = [(1, 0.5), (2, 0.45), (3, 0.47), (4, 0.44), (5, 0.43), (6, 0.42), (7, 0.44), (8, 0.41), (9, 0.42), (10, 0.43)]

print("METHOD 1: Immediate Stop")
best_loss = float('inf')
epochs_ran = 0

for epoch, val_loss in epoch_performances:
    print(f"Epoch {epoch} - Validation Loss: {val_loss}")
    if val_loss < best_loss:
        best_loss = val_loss
        epochs_ran += 1
    else:
        print(f"Stopping early at epoch {epoch} as validation loss increased.")
        break

print(f"Total epochs ran: {epochs_ran}")

print("METHOD 2: Patience")
patience = 2
best_loss = float('inf')
epochs_ran = 0
patience_counter = 0

for epoch, val_loss in epoch_performances:
    print(f"Epoch {epoch} - Validation Loss: {val_loss}")
    if val_loss < best_loss:
        best_loss = val_loss
        patience_counter = 0
        epochs_ran += 1
    else:
        patience_counter += 1
        if patience_counter >= patience:
            print(f"Stopping early at epoch {epoch} as validation loss did not improve for {patience} epochs.")
            break
        epochs_ran += 1

print(f"Total epochs ran: {epochs_ran}")
```

```python
# Scenario:
# ==========
# You have multiple models evaluated over several experiments and want
# to determine which models have the highest average performance. This
# analysis helps identify the most consistently excellent models across
# all experiments and provides insights for model selection decisions.
# ==========
# Task:
# ==========
# Use the model_performance dictionary from the previous exercises to:
# 1. Calculate the average performance for each model across all experiments
# 2. Determine the model(s) with the highest average performance
# 3. Store results in average_performance_dict and best_models list
# Handle cases where multiple models may be tied for best performance.
# ==========
# Solution:
# ==========
print("Quiz 35: Best-Performing Models")

# Assuming model_performance dictionary is already created from the previous exercise

# Create a dictionary to store the total performance and count of evaluations for each model
total_performance = {}
count_evaluations = {}

# Iterate through the model performance dictionary
for experiment, models in model_performance.items():
    for model, performance in models.items():
        if model in total_performance:
            total_performance[model] += performance
            count_evaluations[model] += 1
        else:
            total_performance[model] = performance
            count_evaluations[model] = 1

# Calculate the average performance for each model
average_performance = {model: total_performance[model] / count_evaluations[model] for model in total_performance}

# Find the maximum average performance
max_average_performance = max(average_performance.values())

# Create a list of models with the maximum average performance
best_performing_models = [model for model, performance in average_performance.items() if performance == max_average_performance]

# Print the resulting list
print(best_performing_models)
```
