# **Control Flow**

### **For Loops**

```python
# range (start, stop, step)
# start is inclusive, stop is exclusive
print(list(range(2,6)))
print(list(range(1, 10, 2)))

# will print the range object which not useful
print(range(0,4))

# in order to see the values we can use for loop
for i in range(0, 4):
    print(i)
# or we can convert it to a list
print(list(range(0, 4)))

# Modify list elements using for loop
# Yes, the printed output for the names list will look exactly like it did in the first line. 
# During each iteration, the name variable is set to a string taken from the list. 

# Then the assignment statement creates a new string (name.lower().replace(" ", "_")) and changes the name variable to that string. 
# It doesn't modify the contents of the names list at all. 
# To modify the list you must operate on the list itself, using range, as you saw earlier.
names = ["Joey Tribbiani", "Monica Geller", "Chandler Bing", "Phoebe Buffay"]
for name in names:
    name = name.lower().replace(" ", "_")
print(names)
```

```python
# Building Dictionaries
# Using a for loop to build a dictionary
book_title =  ['great', 'expectations','the', 'adventures', 'of', 'sherlock','holmes','the','great','gasby','hamlet','adventures','of','huckleberry','fin']
word_count = {}
for word in book_title:
    if word not in word_count:
        word_count[word] = 1
    else:
        word_count[word] += 1

print("Word Count:", word_count)

# using .get() method to build a dictionary
# The get(word, 0) method returns the current count for word if it exists in      
# the dictionary, or 0 if it doesn't. Then it adds 1 and stores the result        
# back in word_count[word].
word_count = {}
for word in book_title:
    word_count[word] = word_count.get(word, 0) + 1

print("Word Count:", word_count)
```

```python
# Iterating Through Dictionaries
print ("Iterating Through Dictionaries")
cast = {
           "Jerry Seinfeld": "Jerry Seinfeld",
           "Julia Louis-Dreyfus": "Elaine Benes",
           "Jason Alexander": "George Costanza",
           "Michael Richards": "Cosmo Kramer"
       }

print ("Iterating Through Dictionaries default")
for key in cast:
    print("Actor:", key, "Role:", cast[key])

print ("Iterating Through Dictionaries items()")
# We can iterate also using dict.items() method
for key, value in cast.items():
    print("Actor:", key, "Role:", value)
```

### **Zip and Enum**

```python
# ZIP returns an iterator that combines multiple iterables into one sequence of tuples. 
# each tuple contains the elements in that position from all the iterables
items = ['bananas', 'mattresses', 'dog kennels', 'machine', 'cheeses']
weights = [15, 34, 42, 120, 5]
print(list(zip(items, weights)))

print("list before modifications")
list_x = list(zip(items,weights))
print(list_x)

print("list after modifications")
list_x[0] = ('apples', 10)
print(list_x)
```

```python
# We can do an iteration over zip by the following
print("iteration over zip")
items = ['bananas', 'mattresses', 'dog kennels', 'machine', 'cheeses']
weights = [15, 34, 42, 120, 5]
for item, weight in zip(items, weights):
   print(item, weight)
```

```python
# Unzzzipped List
print("Unzipped List")
manifest = [("bananas", 15), ("mattresses", 34),
("dog kennels", 42), ("machine", 120),
("cheeses", 5)]
items, weights = zip(*manifest)
print("types of items: ", type(items))
print(items)

print("types of weights: ", type(weights))
print(weights)
```

```python
# Enumrate and zip in case used numerical indexing
items = ['bananas', 'mattresses', 'dog kennels', 'machine', 'cheeses']
for i, item in zip(range(len(items)), items):
    print(i, item)

print("Enum without zip")
items = ['bananas', 'mattresses', 'dog kennels', 'machine', 'cheeses']
for i, item in enumerate(items):
    print(i, item)
```

### **List Comprehensions**

```python
cities = ['new york city', 'mountain view', 'chicago', 'los angeles']
capitalized_cities = []
for city in cities:
    capitalized_cities.append(city.title())
print(capitalized_cities)

# OR By using list comprehension
cities = ['new york city', 'mountain view', 'chicago', 'los angeles']
capitalized_cities = [city.title() for city in cities]
print(capitalized_cities)
```

```python
squares = []
for x in range(9):
    squares.append(x**2)
print(squares)

# By using list comprehensions
squares = [x**2 for x in range(9)]
print(squares)
```

```python
# Apply if conditions
squares = [x**2 for x in range(9) if x % 2 == 0]
print(squares)

# Apply if-else conditions
# Note that the if - else is came at the begining
squares = [x**2 if x % 2 == 0 else x + 3 for x in range(9)]
print(squares)
```
