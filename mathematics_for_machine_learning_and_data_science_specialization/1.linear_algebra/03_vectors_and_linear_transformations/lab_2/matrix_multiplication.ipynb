# **Matrix Multiplication**

```python
import numpy as np
```

```python
A = np.array([[4, 9, 9], [9, 1, 6], [9, 2, 3]])
print("Matrix A (3 by 3):\n", A)

B = np.array([[2, 2], [5, 7], [4, 4]])
print("Matrix B (3 by 2):\n", B)
```

```python
# Multiply matrices with matmul Numpy funtion
np.matmul(A, B)
```

```python
# also using dot product @ will work
A @ B
```

```python
# if matrix A is m * n and matrix B is k * p
# then matrix multiplication can happened only in case of 
# n == k other wise matrix multiplication can't be done.

# B * A means (3 * 2)(3 * 3)
# So mutliplication cant be done.
try:
    np.matmul(B, A)
except ValueError as err:
    print(err)

try:
    B @ A
except ValueError as err:
    print(err)
```

## Matrix Convention and Broadcasting


```python
# The following 2 matrices their shape is (3, ) which means a flat list of 3 numbers with no concept of matrices here.
x = np.array([1, -2, -5])
y = np.array([4, 3, -1])
print("Shape of vector x:", x.shape)
print("Number of dimensions of vector x:", x.ndim)

# The result here indicate to a 2D matrix, their size is (3 * 1)
print("Shape of vector x, reshaped to a matrix:", x.reshape((3, 1)).shape)
print("Number of dimensions of vector x, reshaped to a matrix:", x.reshape((3, 1)).ndim)
```

```python
# As long as the concept of matrixes is not exist then the expectation we shall see an error when execute this line
# because we just have 2 (3 * 1) matrices
# but as discussed this is not matrices, so a transpose happened to vector x to be row vector and multiplication happened by y
# y still column vector
print(np.matmul(x,y))


# But in this example we will see failure because we make a force reshape to the vector x to be a matrix
try:
    np.matmul(x.reshape((3, 1)), y.reshape((3, 1)))
except ValueError as err:
    print(err)
```

```python
# As we know dot function is used to get a scalar by multiply 2 vectors by each other
# to understand how these 2 vectors aligned with each other
A = np.array([[4, 9, 9], [9, 1, 6], [9, 2, 3]])
print("Matrix A (3 by 3):\n", A)

B = np.array([[2, 2], [5, 7], [4, 4]])
print("Matrix B (3 by 2):\n", B)

# dot product is not just working with vector, NumPy do a broadcasting this operation to all row and cols in the matrix.
np.dot(A, B)
```

```python
# Normallt Python will broadcast this scalar creating 3 * 3 matrix and perform substraction element by element.
A - 2
```
