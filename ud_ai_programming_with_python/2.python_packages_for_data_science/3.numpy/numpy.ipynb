### **Introduction to NumPy**

```python
# NumPy (Numerical Python) is the fundamental package for scientific computing in Python. It provides:

# Why NumPy?
# Performance: Operations are implemented in C, by precompiled code in C into shared libs and import it to python
# Memory efficiency: NumPy arrays store data contiguously in memory with fixed size but Python lists store pointers to objects scattered throughout memory
# Vectorization: Apply operations to entire arrays without writing loops
# Foundation: Base for other scientific Python packages (pandas, scikit-learn, matplotlib)
```

```python
# This example to comare the speed between using NumPy and Python lists
import time
import numpy as np
x = np.random.random(100000000)

# Case 1 - Normal python APIs
start = time.time()
sum(x) / len(x)
print("Time used by python list = ",time.time() - start)

# Case 2 - NumPy APIs
start = time.time()
np.mean(x)
print("Time used by numpy arrays = ",time.time() - start)
```

### **Built In Functions to create ndarrays**

```python
# Convert list to NumPy array
my_list = [1,2,3,4,5]
x = np.array(my_list)
print(x.dtype)
print(x)
```

```python
# zeros(tuple/int)
x = np.zeros((3,4))

# default dtype = float
print("default array created by zeros:")
print(x)
print(x.dtype)

x = np.zeros((3,4), dtype = int)
print("zeros as int:")
print(x)
```

```python
# ones(tuple/int)
x = np.ones((3,4))

# default dtype = float
print("default array created by ones:")
print(x)

x = np.ones((3,4), dtype = int)
print("ones as int:")
print(x)
```

```python
# full(tuple/int, num to bve filled)
x = np.full((3,4), 5)
print("default array created by full:")
print(x)

print("array elemnts type:", x.dtype)
```

```python
# eye(num): NumPy method to create and identity matrix.
x = np.eye(5)

# element by default float
print(x)

x = np.eye(5, dtype = int)
print(x)
```

```python
# diag([array]): NumPy method to create a matrix with zeros except the diagonal.
x = np.diag([10,20,30,40,50])

print(x)
```

```python
# arange(num): NumPy method create an numpy array with num, note that num is exclusive.
x = np.arange(10)
print("arange(10) = ", x)

# arange(start, stop): NumPy method create an numpy array with start and stop, start inclusive, stop exclusive
x = np.arange(0,5)
print("arange(0,5) = ", x)

# arange(start, stop, step): NumPy method create an numpy array with start and stop, start inclusive, stop exclusive
x = np.arange(1,14,3)
print("arange(1,14,3) = ", x)
```

```python
# linspace(start, stop, n of pts): recomended to be used with non intger values, start and stop inclusive
x = np.linspace(0,25,10)
print(x)

# To exclude last element
x = np.linspace(0,25,10, endpoint=False)
print(x)
```

```python
# reshape(array, new(row, col)): reshape given array to new structure
x = np.arange(20)
print("default array")
print(x)

print("reshape to (4,5) array")
x = np.reshape(x, (4, 5))
print(x)

print("resize to (10,2) array")
x = np.reshape(x, (10, 2))
print(x)
```

```python
# NumPy method, ability to call multiple NumPy method at same line.
# Note at this case reshape() does not take array it self as input
y = np.arange(20).reshape((10,2))
print(y)

x = np.linspace(0,50,10, endpoint=False).reshape(5,2)
print(x)
```

```python
# random(matrix size) function which give us random elements between zero"inclusive" and ones "exclusive"
x = np.random.random((3,3))
print(x)

# randint(start range, stop range, matrix size)
x = np.random.randint(4, 15, (3,3))
print(x)
```

### **Accessing, Deleting, and Inserting Elements Into ndarrays**

```python
import numpy as np
x= np.arange(1,6)
print(x)

# to index array elemenets +ve index
print("x[0] = ", x[0])
print("x[1] = ", x[1])

# to index array elemenets -ve index
print("x[-1] = ", x[-1])
print("x[-5] = ", x[-5])

# to modify arraye element
x[1] = 3
print(x)
```

```python
# How to Access NumPy Arrays
x = np.arange(1,10).reshape(3,3)
print(x)

# to index 2d array
print("x[1, 0] = ", x[1,0])

# to modify an element
x[1, 0] = 7
print(x)

x[0, 1] = 8
print(x)
```

```python
# 1D Array
# delete (numpy_array, [index which need to be removed])
x = np.array([1,2,3,4,5])
print(x)

x = np.delete(x, [0,4])
print(x)
```

```python
# 2D Array
# delete (numpy_array, [index of col / row to be removed based on axis selection], axis)
# axis = 0 --> Row, axis = 1 --> col
y = np.arange(1,10).reshape(3,3)
print(y)

# Remove 1st row
w = np.delete(y, 0, axis=0)
print(w)

# Remove 2nd col
z = np.delete(y, 1, axis=1)
print(z)

# Remove col 1 & 3
v = np.delete(y, [0,2], axis=1)
print(v)
```

```python
# 1D Array
# append(numpy_array, elemnt_value)
x = np.array([1,2,3,4,5])
print(x)

# append one element
x = np.append(x, 6)
print(x)

# append list of element
x = np.append(x, [7,8])
print(x)
```

```python
# 2D Array
# append(numpy_array, [col/row], axis)
# axis = 0 --> Row, axis = 1 --> col
y = np.arange(1,10).reshape(3,3)
print(y)

# add new row
v = np.append(y, [[10,11,12]], axis=0)
print(v)

# add new col
x = np.append(y, [[10],[11],[12]], axis=1)
print(x)
```

```python
# 1D Array
# insert (numpy_array, start_inserting_index, [elements to be inserted])
x = np.array([1,2,5,6,7])
print(x)

x = np.insert(x, 2, [3,4])
print(x)
```

```python
# 2D Array
# insert (numpy_array, start_inserting_index(row/col), [elements to be inserted]/[][][], axis)
x = np.array([[1,2,3],[7,8,9]])
print(x)

# insert new row
x = np.insert(x, 1, [4,5,6], axis=0)
print(x)

# insert new col with multiple valye
x = np.insert(x, 3, [10,11,12], axis=1)
print(x)

# insert new col with single value
x = np.insert(x, 3, 5, axis=1)
print(x)
```

```python
# NumPy arrays stacking
# vstack(arr1, arr2), stack arr1 on top of arr2
x = np.array([1,2])
print(x)

y = np.array([[3,4],[5,6]])
print(y)

z = np.vstack((x,y))
print(z)

# hstack(arr1, arr2), stack 
v = np.hstack((y, x.reshape(2,1)))
print(v)
```

### **Slicing ndarrays**

```python
# We have 3 types to make ndarrays slicing:
# 1.ndarray[start:end]
# 2.ndarray[start:]
# 3.ndarray[:end]
# start inclusive and stop exclusive
import numpy as np
x = np.arange(1,21).reshape(4,5)
print(x)

z = x[1:4, 2:5]
print(z)

v = x[1:, 2:]
print(z)

o = x[:3, 2:]
print(o)

# Grap all rows in col 2, return array rank = 1
y = x[:, 2]
print(y)

# Grap all rows in col 2, return array rank = 2
f = x[:, 2:3]
print(f)
```

```python
# Note, assign numpy array from var to another doesn't mean a copy happened
x = np.arange(1,21).reshape(4,5)

# Assign x to z
z = x[1:, 2:]

# Change value in z
z [0, 0] = 555
print("z: ")
print(z)

# x also affected by the change
print("x: ")
print(x)
```

```python
# In order to make a copy we need to use NumPy "copy" method.
x = np.arange(20).reshape(4,5)
print(x)

# We can use one of the following commands:
z = np.copy(x[1:, 2:])
# z = x[1:, 2:].copy()
print(z)

z[0,0] = 555
print("z: ")
print(z)

# x not affected 
print("x: ")
print(x)
```

```python
# Extract data from NumPy array based on indices
# Extraxt specific Rows and Cols from 
x = np.arange(20).reshape(4,5)
print("x: ")
print(x)

indices = np.array([1,3])

# Note that slicing with using Arrays cosnider to be advanced indexing which NumPy create a copy from the main array no a view as basic indexing
# extract row 1 & 3 from x
y = x[indices, :]
print("y: ")
print(y)

# extract col 1 & 3 from x
z = x[:, indices]
print("z: ")
print(z)
```

```python
# extract the diagonal, extracted array is read only
x = np.arange(20).reshape(4,5)
print("x: ")
print(x)

f = np.diag(x)
print("f contain diagonal elements, read-only:")
print(f)

# Note we can't modify directly in element of f array as it is read-only
# f[0] = 1  compilation error

# In order to modify in f elements:
# - set writable flag to True
# Note this approach you will modifiy in the original numpy array x
f.flags.writeable = True
f[0] = 777
print("'f' after writable flag be true")
print(f)

# return x and f to original values to test the 2nd
f[0] = 0
f.flags.writeable = False

# - create a copy
f = np.diag(x).copy()
f[0] = 555
print("'f' after make a copy of digonal elements of x: ")
print(f)
```

```python
# extract the diagonal, extracted array is read only
x = np.arange(20).reshape(4,5)
print("x: ")
print(x)

# extract elements above the diagonal
k = np.diag(x, k=1)
print("k: ")
print(k)

# extract elements below the diagonal
l = np.diag(x, k=-1)
print("l: ")
print(l)
```

```python
# extract the diagonal, extracted array is read only
x = np.arange(20).reshape(4,5)
print("x: ")
print(x)

# Extract unique elements of the array
x = np.array([[1,2,3], [5,2,8], [1,2,3]])
print("unique elements in x:")
print(np.unique(x))
```

### **Boolean Indexing, Set Operations, and Sorting**

```python
# In previous section we extracted partition from NumPy Array based on incidents as an input from the user.
# Now, we need to filter on array elements with usage of boolean indexing and assign in a new array
x = np.arange(25).reshape(5,5)
print("base x: ")
print(x)

print("'z' array which contian copy of elements which bigger that 10 in x: ")
# z elements can be modified without affecting x
z = x[x>10]
print(z)

y = x[x<=7]
print("'y' which contain elements below 7: ")
print(y)

f = x[(x>10) & (x<17)]
print("'f' which contains elements below than 17 and bigger than 10: ")
print(f)
```

```python
# Update special values in an arrays based on conditions with boolean indexing
x = np.arange(25).reshape(5,5)
print("base x: ")
print(x)

x[(x>10) & (x<17)] = -1
print("update value of elements which their values > 10 & < 17 to -1:")
print(x)
```

```python
# We can use Set operations with NumPy arrays such as (intersection, union, diff)
x = np.array([1,2,3,4,5])
y = np.array([6,7,2,8,4])

print("Intersection(x,y):", np.intersect1d(x,y))
print("Union(x,y):", np.union1d(x,y))
print("Diff(x,y): ", np.setdiff1d(x,y))
```

```python
# Sorting NumPy Arrays
# This line create a numpy array of 10 numbers, min = 1, max = 10, array_size = 10
x = np.random.randint(1,11, size=(10,))
print("base x:")
print(x)

# We have 2 approaches to sort NumPy Array
# - use NumPy sort method which take x as an arg
# Note y contain array x after being sorted but array x does not change
y = np.sort(x)
print("y with sorted x:")
print(y)

print("Note that x didn't changed:")
print(x)

# - use sort method which associated with the NumPy array itself
x.sort()
print("sorted original x:")
print(x)
```

```python
# Now we want to sort 2d arrays
x = np.random.randint(1,11, size=(5,5))
print("base x:")
print(x)

# we can sort by rows:
print("x row sorted:")
print(np.sort(x, axis=0))

print("x col sorted:")
print(np.sort(x, axis=1))
```
