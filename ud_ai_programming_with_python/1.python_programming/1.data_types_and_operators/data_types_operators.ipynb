# **Data types and Operators**

### **Arithmetic Operators**

```python
# Basic arithmetic
print("Addition: 3 + 5 =", 3 + 5)
print("Multiplication: 7 * 3 =", 7 * 3)
print("Division: 10 / 2 =", 10 / 2) # Note: Standard division always results in a float.
```

```python
# Order of Operations (PEMDAS/BODMAS)
# Python respects the mathematical order of operations.
print("Order of Operations: 1 + 2 + 3 * 3 =", 1 + 2 + 3 * 3) # Multiplication is done before addition.
print("Using Parentheses: (1 + 2 + 3) * 3 =", (1 + 2 + 3) * 3) # Parentheses control the order.
```

```python
# Specialized Operators

# Note that Exponentiation is represented by ** in Python.
# WARNING: Do NOT use ^ for exponents; it's a different operator (bitwise XOR).
print("Exponentiation: 3 ** 2 =", 3 ** 2)

# Modulo (%) gives the remainder of a division.
print("Modulus: 9 % 2 =", 9 % 2)

# Integer division (//) divides and then rounds the result DOWN to the nearest whole number.
print("Integer Division: 7 // 2 =", 7 // 2) # 7 / 2 = 3.5, which is rounded down to 3.

# This "rounding down" is important for negative numbers.
print("Negative Integer Division: -7 // 2 =", -7 // 2) # -7 / 2 = -3.5, which is rounded down to -4.
```

### **Variables and Assignment**

```python
# --- 1. What is a Variable? ---
# A variable is a name used to store a value. This allows us to save data
# and refer to it later by its name, making our code more readable and useful.

# this line creates a variable:
mv_population = 74728

# In the line above:
#   - `mv_population` is the variable's name.
#   - `=` is the assignment operator.
#   - `74728` is the value being stored in the variable.

print("The variable 'mv_population' stores the value:", mv_population)
```

```python
# --- 2. The Assignment Operator (=) ---
# A key concept is that `=` in Python is for ASSIGNMENT, not mathematical equality.
# It works from right to left, assigning the value on the right to the name on the left.
# The left side of the `=` must be a variable name.


# --- 3. Using Variables to Access Values ---
# Once a variable is created (defined), you can use its name to access its value.

x = 2
y = x  # Here, 'y' is assigned the value that 'x' currently holds (which is 2).
x = 3

print("After 'y = x', the value of y is:", y) # This will print 2
```

```python
# --- 4. Common Errors: NameError ---
# If you try to use a variable that has NOT been defined yet,
# Python will stop and show a NameError.

# For example, the following line of code would cause a crash because
# the variable 'z' was never created:
# y = z

# The error message would be: NameError: name 'z' is not defined
print("Trying to use an undefined variable like 'z' would cause a NameError.")
```

```python
# --- 5. Multiple Assignment ---
# Python allows you to assign values to multiple variables in a single, concise line.
# This is great for closely related values, like coordinates or dimensions.

# Instead of writing this:
# x = 2
# y = 3
# z = 5

# You can use this convenient shortcut:
x, y, z = 2, 3, 5
print(f"Multiple assignment results: x={x}, y={y}, z={z}")
```

```python
# --- 6. The Importance of Descriptive Names ---
# Good variable names are crucial for writing code that is easy to read and understand.
# While short names like `x` and `y` are fine for simple coordinates, descriptive
# names are better in almost every other case.

# BAD Example (unclear what the variables represent):
# x_bad = 74728
# y_bad = 11.995
# z_bad = x_bad / y_bad

# GOOD Example (clear and self-explanatory):
mv_population = 74728
mv_area = 11.995
mv_density = mv_population / mv_area
print(f"The population density of Mountain View is: {mv_density}")
```

```python
# --- 6. Rules and Conventions for Naming Variables ---

# a) Allowed Characters: There are strict rules for what characters you can use.
#   - Variable names can ONLY contain letters, numbers, and underscores (`_`).
#   - They CANNOT contain spaces.
#   - They MUST start with a letter or an underscore, NOT a number.
#   - ✅ Valid:   `table_6`, `user_name`, `_internal_var`, `number1`
#   - ❌ Invalid: `table 6` (contains a space), `6table` (starts with a number)

# b) Reserved Words & Built-in Identifiers:
#   - You CANNOT use Python's "keywords" (or reserved words) as variable names.
#   - These are words with special meaning, like `if`, `for`, `def`, `class`, etc.
#   - Using them (e.g., `if = 5`) will cause a `SyntaxError`.

# c) The "Pythonic" Style (PEP 8): Use snake_case 🐍
#   - The official style guide for Python recommends "snake_case" for variable names.
#   - This means all words are lowercase and separated by underscores.
#   - ✅ Preferred: `views_per_day`
#   - ❌ Not Preferred: `viewsPerDay` (This is called camelCase and is common in other languages)

pythonic_variable_name = "snake_case_is_best"
print(f"The preferred variable style in Python is: {pythonic_variable_name}")
```

### **Updating Variables & Augmented Assignment**

```python
# --- 1. Reassigning a Variable ---
# You can change the value of a variable at any time by simply assigning
# a new value to it. This overwrites the old value.

mv_population = 74728
print(f"Initial population: {mv_population}")

# The population changes, so we reassign the variable a new value.
mv_population = 78128
print(f"Updated population: {mv_population}")
```

```python
# --- 2. Updating a Variable Based on Its Own Value ---
# A very common task is to modify a variable using its current value.
# For example, if 4000 people move in and 600 move out.

# The "long way" to write this is:
mv_population = 74728 # Reset for the example
mv_population = mv_population + 4000 - 600
print(f"Population after changes (long way): {mv_population}")
```

```python
# --- 3. Augmented Assignment Operators (The Shortcut) ---
# Because the pattern "variable = variable + value" is so common,
# Python provides a cleaner, more concise shortcut called an
# "augmented assignment operator".

# The "short way" using the `+=` operator:
mv_population = 74728 # Reset for the example
mv_population += 4000 - 600 # This is the same as the line above
print(f"Population after changes (short way): {mv_population}")
```

```python
# --- 4. Common Augmented Assignment Operators ---
# This shortcut pattern exists for all standard arithmetic operators.

# Here are the most common ones:
x = 10
print(f"\nInitial value of x: {x}")

x += 2  # Equivalent to: x = x + 2
print(f"After x += 2:  {x}") # x is now 12

x -= 4  # Equivalent to: x = x - 4
print(f"After x -= 4:  {x}") # x is now 8

x *= 3  # Equivalent to: x = x * 3
print(f"After x *= 3:  {x}") # x is now 24

x /= 2  # Equivalent to: x = x / 2
print(f"After x /= 2:  {x}") # x is now 12.0
```

### **Integers and Floats**

```python
# --- 1. Defining `int` and `float` ---

# An `int` is a whole number (without a decimal point).
integer_example = 42

# A `float` is a number that contains a decimal point.
float_example = 42.5

# A key rule: Standard division (`/`) always results in a float, even if the
# result is a whole number.
result = 16 / 4
print(f"The result of 16 / 4 is: {result}") # Output: 4.0
```

```python
# --- 2. Checking Data Types with `type()` ---
# You can use the built-in `type()` function to see the data type of any value.

print(f"The type of 42 is: {type(42)}")       # Output: <class 'int'>
print(f"The type of 42.5 is: {type(42.5)}")   # Output: <class 'float'>
print(f"The type of 4.0 is: {type(4.0)}")     # Output: <class 'float'>
```

```python
# --- 3. Operations and Manual Type Conversion ---
# If you perform an arithmetic operation with an int and a float, the result
# will always be promoted to a float.
mixed_operation = 3 + 2.5
print(f"Result of 3 + 2.5 is: {mixed_operation} (Type: {type(mixed_operation)})")

# You can also manually convert between types:
# a) Converting a float to an int using `int()`
#    This TRUNCATES (cuts off) the decimal part. It does NOT round.
truncated_value = int(49.7)
print(f"int(49.7) becomes: {truncated_value}") # Output: 49

# b) Converting an int to a float using `float()`
#    This simply adds a .0 to the end of the number.
float_value = float(50)
print(f"float(50) becomes: {float_value}") # Output: 50.0
```

```python
# --- 4. A Word of Caution: Floating-Point Imprecision ---
# It's important to know that floats are stored as approximations in computer
# memory. This can sometimes lead to results that are not perfectly precise.

# For example, 0.1 + 0.1 + 0.1 is not exactly 0.3 in Python.
imprecision_example = 0.1 + 0.1 + 0.1
print(f"The result of 0.1 + 0.1 + 0.1 is: {imprecision_example}")
# Expected output: 0.30000000000000004

# For most applications, this tiny difference doesn't matter, but it's a
# fundamental concept in computer science to be aware of.
```

### **Booleans, Comparison, and Logical Operators**

```python
# --- 1. The `bool` Data Type ---
# A Boolean (`bool`) is a data type that can only have one of two values:
# True or False (note the capitalization).
# 
is_weekend = True
is_sunny = False
print(f"The type of 'is_weekend' is: {type(is_weekend)}") # Output: <class 'bool'>
```

```python
# --- 2. Comparison Operators ---
# Booleans are most often created by using comparison operators to compare values.
# These expressions always evaluate to either `True` or `False`.

age = 25

# Greater than / Less than
print(f"Is age > 18? {age > 18}")     # Output: True
print(f"Is age < 21? {age < 21}")     # Output: False

# Greater than or equal to / Less than or equal to
print(f"Is age >= 25? {age >= 25}")   # Output: True
print(f"Is age <= 25? {age <= 25}")   # Output: True

# Equal to
# ⚠️ IMPORTANT: Use a double equals sign `==` for comparison.
# A single equals sign `=` is for assignment.
print(f"Is age == 25? {age == 25}")   # Output: True

# Not equal to
print(f"Is age != 30? {age != 30}")   # Output: True
```

```python
# --- 3. Logical Operators ---
# Logical operators are used to combine or modify Boolean expressions.

# a) `and`
#    Returns `True` only if BOTH sides are true.
#    Example from the video: checking if someone is a teenager.
teen_age = 14
is_teen = teen_age > 12 and teen_age < 20
print(f"\nIs a 14-year-old a teen? {is_teen}") # True and True -> True

# b) `or`
#    Returns `True` if AT LEAST ONE side is true.
can_enter_club = age > 21 or age == 21 # (is age > 21 OR is age == 21?)
print(f"Can a 25-year-old enter the club? {can_enter_club}") # False or False -> False
# Let's try another example for 'or'
is_day_off = is_weekend or not is_sunny
print(f"Is it a day off? {is_day_off}") # True or True -> True

# c) `not`
#    Inverts the Boolean value (True becomes False, False becomes True).
is_adult = age >= 18
is_child = not is_adult
print(f"Is a 25-year-old a child? {is_child}") # not True -> False
```

### **Strings**

```python
# --- 1. Creating Strings ---
# You can create strings using either single quotes (') or double quotes (").
# They work in exactly the same way.

string_1 = "Hello, world!"
string_2 = 'Hello, world!'
print(f"String with double quotes: {string_1}")
print(f"String with single quotes: {string_2}")
```

```python
# --- 2. Handling Quotes Inside Strings ---
# Sometimes you need to include a quote character inside your string.

# a) The easy way: Use the opposite type of quote to define the string.
#    If your string contains single quotes, define it with double quotes.
my_string = "It's a beautiful day."
print(my_string)

#    If your string contains double quotes, define it with single quotes.
my_string_2 = 'He said, "Hello!"'
print(my_string_2)


# b) The flexible way: Use the escape character (`\`).
#    A backslash tells Python to treat the next character as a literal part
#    of the string, not as a special character. This is necessary if your
#    string contains BOTH single and double quotes.
#    (Here are your examples, integrated and explained)
print("Single quote with escape character: \'")
print("Double quote with escape character: \"")
sales_pitch = "The salesman said, \"I think you\'re great!\""
print(sales_pitch)
```

```python
# --- 3. String Operators ---
# Some arithmetic operators work on strings, but have a different meaning.

# a) Concatenation (`+`): Combines or joins strings together.
#    (Your example)
print("Sum String: Hello + World! =", "Hello" + " " + "World!") # Note the explicit space

# b) Repetition (`*`): Repeats a string a given number of times.
#    (Your example)
print("Repeat String: 'Hello' * 3 =", "Hello" * 3)

# c) Unsupported operators (like `/`, `-`) will cause a TypeError.
#    "Hello" / 2  # This would raise a TypeError.
```

```python
# --- 4. The `len()` Function ---
# `len()` is a very useful built-in function that returns the length of a string
# (i.e., the number of characters it contains).
# (Your example)
print("Length of 'Ahmed':", len("Ahmed")) # Output: 5

# Unlike print(), len() returns a value that can be stored in a variable.
name_length = len("Ahmed")
print(f"The length is stored in a variable: {name_length}")
```

### **Types and Type Conversion**

```python
# --- 1. Recap of Basic Data Types ---
# So far, we've worked with four main data types:
# - `int`: For whole numbers (e.g., 10, -50).
# - `float`: For numbers with decimals (e.g., 3.14, -0.5).
# - `bool`: For logical values (True, False).
# - `str`: For text (e.g., "Hello", 'Python').
```

```python
# --- 2. Checking a Value's Type with `type()` ---
# The built-in `type()` function is used to find the data type of any variable or value.
# As the video showed, the same number can be represented by different types.
print(f"The type of 633 is: {type(633)}")
print(f"The type of '633' is: {type('633')}")
print(f"The type of 633.0 is: {type(633.0)}")
# (This is your example, which works perfectly)
print("Type of 'Hello':", type("Hello"))
```

```python
# --- 3. Type Conversion (Casting) ---
# It's often necessary to convert a value from one type to another. This is
# known as "type conversion" or "casting." We do this using functions that
# are named after the type we want to convert to: `int()`, `float()`, `str()`.

# a) Converting to a String with `str()`
#    This is essential when you want to combine text with a number. You cannot
#    use the `+` operator to add a string and a number directly.
house_number = 13
full_address = str(house_number) + " The Crescent, Belmont"
print("Casting 13 to string: " + full_address)

# b) Converting a String to a Number with `int()` or `float()`
#    This is useful when you have a number that is currently stored as text.
grams = "35.0"
print("\nInitial 'grams' type:", type(grams))
grams = float(grams)
print("After float(grams), the type is:", type(grams))
print(f"Now we can do math with it: {grams / 2}")

# c) Converting a Float to an Int with `int()`
#    This is useful for getting just the whole number part of a float.
#    Remember, this TRUNCATES (cuts off) the decimal; it does not round.
count_float = 4.8
count_int = int(count_float)
print(f"\nint({count_float}) becomes: {count_int}")
```

### **String Methods**

```python
# A method is a function associated with a particular data type. You can
# think of them as special actions that an object knows how to perform on itself.
# They are always called using dot notation: `object.method_name()`.

# --- Common String Methods (from the video) ---

# .title(): Returns a new string where the first letter of each word is capitalized.
print("title() method:", "sebastian thrun".title())

# .islower(): Checks if all cased characters in the string are lowercase.
# It returns a boolean (True or False).
print("islower() method:", "sebastian thrun".islower())

# .count(key): Returns the number of times a specific substring (the key) appears
# within the string. It returns an integer.
print("count() method:", "One fish, two fish, red fish, blue fish".count("fish"))
```

```python
# .split(): powerful tool for breaking a larger string
# into a list of smaller strings.

# The full syntax is: string.split(separator, maxsplit)
# - `separator`: This is the character (or string) where the split should happen.
#                It's optional; if you don't provide it, Python splits by any whitespace.
# - `maxsplit`: This is the maximum number of splits to perform.
#               It's also optional. The rest of the string is returned as the final item.

new_str = "The cow jumped over the moon."
# 1. Basic usage of .split()
# When no arguments are given, it splits the string at every whitespace character.
print("Basic usage of split() method:")
print(new_str.split())
# Output: ['The', 'cow', 'jumped', 'over', 'the', 'moon.']

# 2. Using a separator and maxsplit
# Here, the separator is explicitly a space " ", and it will only split a maximum of 3 times.
print("\nSeparator is space, and the maxsplit argument is set to 3:")
print(new_str.split(" ", 3))
# Output: ['The', 'cow', 'jumped', 'over the moon.']

# 3. Using a different separator
# This example splits the string only where it finds a period character.
print("\nUsing '.' or period as a separator:")
print(new_str.split('.'))
# Output: ['The cow jumped over the moon', '']
# Note: The empty string '' appears because the separator was at the very end.

# 4. Using maxsplit without a specific separator
# Passing `None` as the separator is the same as the default behavior
# (splitting by any whitespace), but here we limit it to 3 splits.
print("\nUsing no separators but having a maxsplit argument of 3:")
print(new_str.split(None, 3))
# Output: ['The', 'cow', 'jumped', 'over the moon.']
```
